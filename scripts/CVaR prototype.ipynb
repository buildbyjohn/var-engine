```python
import numpy as np

# --- Inputs (we'll use placeholder numbers for now, real BNM data comes in Phase 2) ---
S0 = 4.20          # starting USD/MYR price
mu = 0.02           # annualised drift (placeholder guess)
sigma = 0.08        # annualised volatility (placeholder guess)
T = 10/252          # horizon: 10 trading days, as a fraction of a year
n_paths = 100_000
V0 = 1_000_000      # notional position value in RM

np.random.seed(42)  # fixed seed for reproducibility

# --- Step 1: simulate ending prices ---
Z = np.random.standard_normal(n_paths)
S_T = S0 * np.exp(((mu - sigma**2 / 2) * T) + (sigma * np.sqrt(T) * Z))   # <-- fill in the GBM exponent using mu, sigma, T, Z

# --- Step 2: convert to portfolio losses ---
V_T = V0 * (S_T / S0)             # <-- scale V0 by the price ratio S_T/S0
losses = V0 - V_T             # <-- V0 minus V_T

# --- Step 3: VaR ---
VaR_99 = np.percentile(losses, 99)

# --- Step 4: CVaR ---
tail_losses = losses[losses >= VaR_99]
CVaR_99 = tail_losses.mean()

print(f"99% VaR: {VaR_99:,.2f}")
print(f"99% CVaR: {CVaR_99:,.2f}")
```
